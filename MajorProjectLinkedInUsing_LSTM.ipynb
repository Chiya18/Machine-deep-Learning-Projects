```python
import pandas as pd
import re
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.preprocessing import LabelEncoder
import numpy as np

# Example: Load your dataset (CSV with 'post' and 'label')
df = pd.read_csv("/content/linkedinFauxPostDetection_Dataset - Sheet1 (1).csv")
print(f"DataFrame shape after loading: {df.shape}")

# Step 1.1: Clean text (remove links, symbols, etc.)
def clean_text(text):
    text = re.sub(r"http\S+", "", text)   # remove URLs
    text = re.sub(r"[^a-zA-Z]", " ", text) # keep only letters
    text = text.lower().strip()           # lowercase & trim
    return text

# Convert the 'Data' column to string type, coercing errors
df["clean_post"] = df["Data"].astype(str).apply(clean_text)
print(f"DataFrame shape after cleaning text: {df.shape}")

# Step 1.2: Tokenization
tokenizer = Tokenizer(num_words=5000)  # keep top 5000 words
tokenizer.fit_on_texts(df["clean_post"])

# Convert posts to sequences of numbers
X = tokenizer.texts_to_sequences(df["clean_post"])
print(f"Shape of X after tokenization: {len(X)}")


# Step 1.3: Padding (make all sequences same length)
X = pad_sequences(X, maxlen=100)  # truncate/pad to 100 words
print(f"Shape of X after padding: {X.shape}")


# Labels
y = df["Label (Fake / Real )"].values

# Encode labels to numerical values (0 and 1)
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)
print(f"Shape of y after label encoding: {y.shape}")
print(f"Unique values in y after encoding: {np.unique(y)}")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

# Step 2.1: Define the model
model = Sequential()

# Embedding layer converts word indices into word vectors
model.add(Embedding(input_dim=5000, output_dim=128, input_length=100))

# LSTM layer
model.add(LSTM(128, dropout=0.2, recurrent_dropout=0.2))

# Dense output layer (binary classification: fake or real)
model.add(Dense(1, activation='sigmoid'))

# Step 2.2: Compile
model.compile(loss='binary_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

# Step 2.3: Train
model.fit(X, y, epochs=5, batch_size=64, validation_split=0.2)
```

```python
import numpy as np

def predict_post(text, tokenizer, model, maxlen=100):
    # Clean text (same as training)
    def clean_text(t):
        import re
        t = re.sub(r"http\S+", "", t)
        t = re.sub(r"[^a-zA-Z]", " ", t)
        return t.lower().strip()

    # Preprocess
    seq = tokenizer.texts_to_sequences([clean_text(text)])
    pad = pad_sequences(seq, maxlen=maxlen)

    # Predict
    prob = model.predict(pad)[0][0]

    # Print results
    print(f"🔎 Raw probability: {prob:.4f}")
    if prob > 0.5:
        print("✅ Model says: Fake (with confidence {:.2f}%)".format(prob*100))
    else:
        print("✅ Model says: Real (with confidence {:.2f}%)".format((1-prob)*100))

# Example usage
test_post = "Breaking news: Earn $5000 a day by working from home!!!"
predict_post(test_post, tokenizer, model)
```
